This notebook inspects remote sensing benchmarks (RSVQA, BigEarthNet-MM, and CDVQA) and verifies directory structure and telemetry tags.

```python
import os
import sys
import json
import numpy as np
import matplotlib.pyplot as plt

# Ensure repository root is on path
repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if repo_root not in sys.path:
    sys.path.append(repo_root)

print(f"SatQuery AI Root: {repo_root}")
data_dir = os.path.join(repo_root, "data")
print(f"Data Directory exists: {os.path.exists(data_dir)}")
```

```python
# Check for BigEarthNet-MM or mounted datasets
bigearth_dir = os.path.join(data_dir, "BigEarthNet-MM")
if os.path.exists(bigearth_dir):
    manifest_path = os.path.join(bigearth_dir, "train_manifest.json")
    if os.path.isfile(manifest_path):
        with open(manifest_path, "r") as f:
            samples = json.load(f)
        print(f"Mounted BigEarthNet-MM samples: {len(samples)}")
    else:
        print("BigEarthNet-MM directory exists but train_manifest.json is missing.")
else:
    print("BigEarthNet-MM dataset not mounted. Note: SatQuery AI never fabricates missing datasets.")
```

```python
# Inspect synthetic demonstration rasters in data/samples
samples_dir = os.path.join(data_dir, "samples")
if os.path.exists(samples_dir):
    sample_files = [f for f in os.listdir(samples_dir) if f.endswith('.tif')]
    print(f"Synthetic test rasters available in data/samples: {sample_files}")
else:
    print("No sample rasters found.")
```
