Defines the adapter architecture and projection modules for adapting foundation vision-language models to multi-modal satellite Earth observation inputs.

```python
import torch
import torch.nn as nn

class MultimodalProjectionAdapter(nn.Module):
    """
    Projects fused multi-band satellite raster embeddings into the text embedding space
    of a foundation language model.
    """
    def __init__(self, visual_dim: int = 768, hidden_dim: int = 2048, text_dim: int = 3584):
        super().__init__()
        self.proj = nn.Sequential(
            nn.Linear(visual_dim, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, text_dim)
        )

    def forward(self, visual_tokens):
        return self.proj(visual_tokens)

adapter = MultimodalProjectionAdapter()
print(adapter)
num_params = sum(p.numel() for p in adapter.parameters())
print(f"Adapter trainable parameter count: {num_params:,}")
```
